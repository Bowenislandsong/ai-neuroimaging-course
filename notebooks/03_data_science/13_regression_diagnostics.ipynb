## Understand the transformation

A regression coefficient is interpretable only with the design columns, units, and modeling assumptions. Correlated predictors make separate slopes sensitive to small perturbations. A model can give stable predictions inside the observed range while the coefficients individually change a great deal. This is especially relevant when multiple imaging features carry nearly the same information.

Residual plots examine what the fitted relationship fails to describe. A curved pattern suggests a missing nonlinear relationship; a changing spread suggests unequal variance; serial structure suggests dependence. None of these plots proves every assumption when it looks unremarkable. They are targeted opportunities to falsify a proposed model. Influential observations should be investigated, not deleted automatically.

We construct two almost identical predictors. Tiny outcome perturbations cause larger shifts in individual coefficients than in fitted outcomes. Compare this with a design using only one of the redundant predictors. Regularization can stabilize estimation, but it introduces a penalty and changes the bias-variance tradeoff; it is not a free restoration of separate biological effects. The modeling strand examines ridge and validation in detail.

## Experiment

Compare two outcome vectors differing by small noise under a collinear design. Deliberate error: rank brain regions using unstable coefficient magnitudes as if they were independent causal importance.

Run the following cells in order. Before each, predict what should remain unchanged and what should differ. The assertions test specific mathematical or bookkeeping properties, not clinical validity.

```python
import numpy as np
rng=np.random.default_rng(313); x=rng.normal(size=100)
x2=x+rng.normal(0,.0001,100)
X=np.column_stack([np.ones(100),x,x2]); y=2+3*x+rng.normal(0,.02,100)
y2=y+rng.normal(0,.002,100)
b1=np.linalg.lstsq(X,y,rcond=None)[0]; b2=np.linalg.lstsq(X,y2,rcond=None)[0]
coef_change=np.linalg.norm(b2-b1); prediction_change=np.linalg.norm(X@(b2-b1))
assert np.linalg.cond(X)>1000
assert coef_change>prediction_change
print('Condition number:',np.linalg.cond(X))
print('Coefficients:',b1,b2)
print('Coefficient/prediction L2 change:',coef_change,prediction_change)
```
