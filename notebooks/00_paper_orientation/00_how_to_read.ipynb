## Start without AI

Choose one opening paper. Read its title and abstract yourself. Complete these sentences before asking for help:

1. The scientific or practical problem seems to be …
2. The authors want to change or establish …
3. The evidence I would need to believe that is …
4. I do not yet understand …

An uncertain first attempt is useful. A polished summary generated before you read the source gives the instructor little evidence of your understanding.


## Practice on invented claims

These statements are **course-created examples, not quotations or results from any assigned paper**:

- “We tested on 120 scans from 30 people, so we tested 120 independent people.”
- “The model has the best average score, so it must be the best choice for every hospital.”
- “The highlighted brain region explains why the model made its prediction.”

For each, identify what additional information you would seek. Do not try to calculate a p-value. Consider the observation unit, intended use, comparator, and difference between a visualization and a tested explanation. Rewrite each statement so it says only what its provided evidence establishes.

## Ask the AI for a limited kind of help

> I am reading this paper for motivation before studying its mathematics. I will provide the abstract and one figure caption. First ask what I think the problem and evidence are. Then explain only the terms needed to interpret that evidence. Cite a section or figure for each paper-specific assertion. Separate the authors' claim, my interpretation, and what remains unknown. Do not invent results, give me a whole-paper summary from memory, or start writing a pipeline. Ask which fundamental concept I want to understand next.

If the tutor has no document access, supply the relevant source excerpt yourself. “The model knows the paper” is not a source check. After the explanation, close it and restate the result in your own words. Return to the original figure to check your restatement.
