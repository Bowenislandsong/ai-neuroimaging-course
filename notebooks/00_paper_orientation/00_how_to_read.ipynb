### Seminar Architecture & Pedagogical Goal

In graduate neuroimaging seminars at institutions such as MIT, Stanford, and the USC Neuroimaging and Informatics Institute (NIIN), students do not begin by memorizing software flags or copying deep-learning architectures. They begin by auditing **how raw physical measurements are transformed into scientific claims**—and where that chain of evidence breaks. Modern neuroimaging sits at the intersection of magnetic resonance physics, neurovascular physiology, high-dimensional spatial processing, experimental design, and statistical or self-supervised machine learning. When an analyst or an AI coding assistant writes a ten-line Python snippet or invokes a pretrained foundation model, that code silently commits to dozens of scientific assumptions about coordinates, noise distributions, sampling units, and generalization boundaries.

This opening orientation (`R00`–`R03`) establishes your **supervisory habits** before you execute a single line of code in `F01`. Rather than treating published papers as infallible textbooks or treating recent foundation models as universal winners, you will learn to read every paper as an **engineered argument under constraints**. Specifically, you will separate:
1. **The Scientific Estimand or Engineering Target:** What biological quantity, anatomical correspondence, or out-of-sample prediction is the paper trying to establish?
2. **The Observation & Generalization Units:** Does one data point represent a $2\text{ mm}^3$ voxel, a $700\text{ ms}$ TR volume, a 15-minute scanning run, a longitudinal session, an individual participant, an acquisition scanner/site, or an entire analysis team?
3. **The Transformation & Evaluation Contract:** What operations stand between the scanner k-space reconstruction and the plotted figure panel, what baseline comparator is held fixed, and which failure modes remain invisible to the headline metric?

## The Three-Pass Method Adapted for Neuroimaging & AI

In Keshav's classic Three-Pass Method, readers move from a bird's-eye survey to figure comprehension to virtual re-implementation. In our [three-pass reading guide](../../curriculum/papers/READING_METHOD.md), we adapt this protocol specifically for empirical neuroimaging and modern machine-learning manuscripts:

### Pass 1: Motivation, Estimand, and Provenance Gate (15–20 minutes)
- **What to read:** Title, Abstract, Introduction, section headings, the exact manuscript version/date/license metadata, and the Discussion/Limitations section. **Skip all equations and architecture diagrams on Pass 1.**
- **What to extract:**
  1. *The Bottleneck:* What failure mode in existing neuroimaging practice is the paper addressing (e.g., hidden pipeline variability in `PP01`, small-sample effect inflation in `PD02`, fixed-parcellation information loss in `PM05`)?
  2. *The Estimand / Task Type:* Is the paper performing **image transformation** (mapping coordinates/intensities), **population parameter estimation** (inferring a group mean contrast $\mu$ or correlation $r$), **out-of-sample individual prediction** (predicting age, diagnosis, or survival $\hat{y}_i$ for an unseen person), or **meta-analytical auditing** (measuring variance across analysts or benchmarks)?
  3. *The Provenance Record:* Are you reading a peer-reviewed journal version with published corrections (e.g., `PD02` 2022 correction; `PD05` August 2016 PNAS correction) or a specific timestamped arXiv preprint (`PP04` v3 `2405.14019v3`; `PM05` v1 `2601.23090v1`)? Never cite a moving target without locking the version.

### Pass 2: Forensic Figure & Table Audit (40–60 minutes — Your Opening Seminar Target)
- **What to read:** The assigned Figure panels, captions, and Table rows, cross-referencing only the specific Methods paragraphs needed to define the plotted axes and units.
- **The Six-Point Forensic Figure Checklist:**
  1. **Axis & Metric Semantics:** What is the exact quantity on the $x$-axis and $y$-axis? What are the physical or statistical units (e.g., millimeters, Dice coefficient $\in [0,1]$, Pearson $r$, Mean Absolute Error in years, Harrell's C-index)? Which direction indicates improvement ($\uparrow$ vs. $\downarrow$)?
  2. **The Observation Unit & Denominator ($N$):** When a boxplot, violin plot, or table cell reports a summary number, what is one observation in the denominator? Is $N$ the number of voxels, 2D slices, scanning runs, unique human participants, clinical hospital sites, random seeds, or analysis teams (`PD01`: 70 reporting teams in Fig. 1 vs. 64 usable unthresholded maps in Fig. 2)?
  3. **Error Bars & Uncertainty Display:** What do the shaded bands or whiskers represent—standard deviation (spread of the data), standard error of the mean (precision of the mean), a $95\%$ parametric confidence interval, or non-parametric participant bootstrap quantiles? Over what resampling unit were they computed?
  4. **The Comparator & Fairness Contract:** What baseline is being compared? Did the baseline receive the same hyperparameter tuning budget, input resolution, skull-stripping preprocessing, and training data split (`PM04`)?
  5. **The Information Boundary (Hold-Out Split):** What data were used to pretrain representations, select features, harmonize scanners, or tune hyperparameters, and what data were strictly isolated for evaluation? Were repeated scans from the same person kept together? Was the test set drawn from an unseen scanner/hospital site (`P03`)?
  6. **Exception Hunting:** Where does the proposed method *fail* or underperform a simpler baseline (e.g., `PM05` Table 3 on the SALD age-regression benchmark; `PP04` 3D Thin-Plate Spline registration vs. ANTs SyN and SynthMorph at low initial misalignment on skull-stripped scans)?

### Pass 3: Mathematical & Computational Reconstruction (Weeks 3–26)
- **What to read:** Full Methods, equations, supplementary derivations, and official code repositories—undertaken **after** you complete the relevant foundational notebooks (`F01`–`F04`, `PR`, `D`, `DS`, `M`).
- **What to execute:** In Assignments `A2`, `A3`, and `A4`, you will isolate one core mathematical transformation from the paper, implement a controlled synthetic or public-data experiment, deliberately break one assumption, and update your evidence ledger.

## Forensic Practice: Auditing Invented & AI-Inflated Claims

Below are six statements that represent **common AI-generated reading mistakes and press-release overgeneralizations**. They are **course-created pedagogical examples, not quotations from any assigned paper**:

### Case Set A: Core Observation-Unit, Generalization, and Mechanism Confusions
1. **Claim 1 (Pseudoreplication):** *“We evaluated our biomarker on 120 resting-state scans acquired across 30 participants (4 sessions each), achieving $p < 0.001$ across $N = 120$ independent people.”*
2. **Claim 2 (Aggregation & Site Shift Fallacy):** *“Our 3D foundation model achieved the highest average score across five public benchmarks, proving it is the optimal clinical choice for every hospital scanner.”*
3. **Claim 3 (Attribution vs. Causality Fallacy):** *“The transformer attention map highlights the bilateral amygdala, which explains the neurobiological mechanism causing the patient's symptoms.”*

### Case Set B: Paper-Specific Misreadings Commonly Produced by Unsupervised LLMs
4. **Claim 4 (Workflow vs. Component Confusion — `R01`):** *“Because BrainMorph (`PP04`) registers brain MRIs in under a second using learned keypoints, it completely replaces fMRIPrep (`PP01`) for functional MRI preprocessing.”*
5. **Claim 5 (Estimand & Sample-Size Confusion — `R02`):** *“Marek et al. (`PD02`) proved that every fMRI study with fewer than 2,000 subjects is scientifically invalid, whereas BrainIAC (`PM03`) proved that 50 subjects are always sufficient for any brain study.”*
6. **Claim 6 (Terminology & Benchmark Exception Blindness — `R03`):** *“Omni-fMRI (`PM05`) is an atlas-free foundation model that operates directly on raw scanner data without preprocessing and outperforms every competing method on every benchmark in Table 3.”*

### How to Conduct a Claim Audit (Before & After Comparison)

See how a passive reader accepts an AI summary versus how a rigorous neuroimaging supervisor rewrites it:

| Dimension | Unsupervised AI / Passive Summary (Reject) | Supervised Forensic Rewrite (Target Standard) |
|---|---|---|
| **Sample Unit** | Treats 120 longitudinal scans or 50,000 2D slices as $N$ independent human subjects. | Explicitly separates $N_{\text{people}} = 30$ (unit of population inference) from $k = 4$ repeated sessions per person (unit of within-subject reliability). |
| **Scope of Claim** | Turns a mean benchmark table win into universal clinical superiority across unseen scanners. | Bounds the claim to the tested acquisition protocols, target labels, and adaptation regime; inspects site-specific variance and table exceptions. |
| **Mechanism vs. Association** | Treats a regression weight, Grad-CAM heatmap, or self-attention weight as a causal lesion mechanism. | Identifies the map as a descriptive statistical or predictive feature weight conditioned on the model class, preprocessing, and nuisance covariates. |

### Detailed Grading & Discussion Notes for Seminar Leaders

1. **Audit of Claim 1 (120 scans from 30 participants):**
   - *Core flaw:* **Hierarchical pseudoreplication** (confusing within-person longitudinal/session repeats $k=4$ with independent population draws $N=30$).
   - *Why it matters:* Two resting-state or task scans from the same brain share anatomical folding, vascular architecture, head-motion habits, and baseline trait variance. Treating all 120 scans as independent degrees of freedom ($df = 118$ instead of $df = 28$) underestimates standard errors and severely inflates Type I false-positive rates—or, in machine learning, leaks subject identity across train and test folds (`P03`, `D02`, `DS07`).
   - *Acceptable student rewrite:* "We evaluated our measure on 120 scans collected across 30 individuals (4 scans per person); establishing generalization to new people requires aggregating within person or using participant-grouped cross-validation with $N=30$ as the unit of independence."

2. **Audit of Claim 2 (Best average score across benchmarks = best for every hospital):**
   - *Core flaw:* **Ecological / aggregation fallacy and unverified domain shift** (`PM03`, `PM04`, `PD07`).
   - *Why it matters:* An unweighted mean across tasks or datasets hides high variance, opposite metric directions, and catastrophic failure on specific sub-tasks or acquisition regimes (e.g., 1.5T clinical thick-slice FLAIR vs. 3T research isotropic MPRAGE, or demographic shift). Furthermore, hospital deployment requires local calibration, computational feasibility, and subgroup fairness checks (`M04`, `M18`, `P03`).
   - *Acceptable student rewrite:* "The model achieved the highest mean score across the five evaluated benchmark tasks under the reported adaptation protocol; however, task-specific variance, scanner batch shifts, and local clinical calibration must be tested before selecting it for a new hospital."

3. **Audit of Claim 3 (Highlighted attention/saliency region = causal biological mechanism):**
   - *Core flaw:* **Reverse inference and conflation of predictive attribution with causal mechanism** (`PM01`, `PM05`, `M17`, `M18`).
   - *Why it matters:* A high weight or attention score in a multivariate model indicates that voxel intensities in that region contributed to the model's mathematical output *given the other features and preprocessing*—which can easily reflect correlated head motion, physiological pulsatility, CSF partial volume, or vascular drainage rather than neural causation (`PB01`, `PP02`). Establishing a causal mechanism requires interventional perturbation, lesion evidence, or biophysically validated forward models.
   - *Acceptable student rewrite:* "The model assigned high attribution weight to voxels overlapping the amygdala parcel in this predictive task; whether this reflects neural activity, correlated physiological noise, or a causal disease mechanism requires independent confound control and experimental validation."

4. **Pedagogical Note on Technical Vocabulary:**
   - Do **not** penalize students in Weeks 1–2 if they do not yet know the formal terms *intraclass correlation coefficient (ICC)*, *Linear Mixed-Effects (LME)*, *ComBat empirical Bayes*, or *Benjamini-Hochberg FDR*. Reward them when they notice the *structural gap* in the evidence and formulate the exact question that those methods will answer in `F01`–`P04`.
