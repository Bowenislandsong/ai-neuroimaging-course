## A small reviewer meeting

Spend 15 minutes as the paper's advocate: explain the problem and strongest inspected evidence. Spend 15 minutes as a reviewer: challenge the split, baseline, metric or interpretation using a specific source location. Then write a balanced 200-word decision that preserves both the contribution and its boundary.

If studying alone, record the advocate explanation first, then ask the AI for three possible objections and check each against the paper. Do not reward an objection for sounding skeptical; it must identify an actual uncertainty or a testable alternative. A source-grounded favorable conclusion is as acceptable as a source-grounded concern.

**Deliver:** the figure/table evidence rows, three rewritten claims, one four-box diagram, the reviewer decision and a first [SOTA comparison card](../../curriculum/coursework/SOTA_AUDIT.md). A precise table exception is useful evidence; it does not by itself erase the paper's other contributions.
