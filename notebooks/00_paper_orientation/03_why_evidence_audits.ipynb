### Seminar Framing: Analytical Vibration ("The Multiverse") and Frontier Benchmark Audits

In `R01`, you saw that a raw scan undergoes a complex chain of spatial and temporal preprocessing steps before analysis. In `R02`, you saw that sample sizes, phenotype SNR, and train/test boundaries govern whether a result generalizes to a new person. Now, in `R03`, we confront the final supervisory question before entering the computational labs: **How do researcher degrees of freedom and benchmark design choices shape whether a result looks "significant" or "state-of-the-art"?**

We pair two landmark studies that every modern neuroimaging researcher must know how to read side by side:
1. **The Analytical Multiverse Audit (`PD01`):** Botvinik-Nezer et al. (*NARPS*, Nature, 2020), where 70 independent research teams tested the exact same 9 neuroimaging hypotheses on the exact same $N = 108$ task-fMRI dataset—yielding wide variability in binary "yes/no" significance decisions alongside striking convergence in underlying unthresholded effect maps (connecting directly to [Eklund et al., 2016, `PD05`](../../curriculum/papers/design.md#pd05) on cluster inference).
2. **The Frontier fMRI Foundation Model (`PM05`):** Wang et al. (*Omni-fMRI*, ICML 2026 / assigned arXiv v1 `2601.23090v1`), which proposes an "atlas-free" fMRI foundation model using activity-aware spatiotemporal SLIC patching—providing a live 2026 test case for auditing tokenization assumptions, preprocessing dependencies, and table counterexamples (connecting to [OpenMind, Wald et al., 2025, `PM04`](../../curriculum/papers/modeling.md#pm04)).

### Side-by-Side Forensic Comparison (`PD01` vs. `PM05` & Audit Strand Context)

| Analytical Dimension | `PD01` · *NARPS* (Botvinik-Nezer et al., 2020) | `PM05` · *Omni-fMRI* (Wang et al., 2026, arXiv v1) | Connections to `PD05` (Eklund) & `PM04` (OpenMind) |
|---|---|---|---|
| **Core Problem Addressed** | **Analytical flexibility ("researcher degrees of freedom"):** How much do reasonable, expert-chosen fMRI preprocessing and statistical modeling pipelines diverge when testing identical biological hypotheses on identical raw data? | **Representation bottleneck of fixed brain atlases:** Standard fMRI models project 4D voxel data onto fixed anatomical parcellations (e.g., AAL, Schaefer-400), discarding fine voxel geometry and imposing mismatched boundaries across individuals. | [`PD05` (Eklund et al., 2016)](../../curriculum/papers/design.md#pd05) showed how parametric cluster-extent thresholding choices (`p < 0.01` vs `p < 0.001`) drastically alter false-positive rates—one of the primary drivers of NARPS divergence! |
| **Input Data & Preprocessing Dependency** | Raw BIDS task-fMRI dataset ($N = 108$ participants performing a mixed-gambles loss-aversion task); half the teams used standardized **fMRIPrep (`PP01`)** derivatives while half ran custom FSL/SPM/AFNI pipelines. | Preprocessed 4D MNI-space fMRI volumes across 49,561 runs from multiple cohorts—**"atlas-free" means no fixed ROI parcellation is applied, NOT that raw unaligned scanner data are fed in without `PP01`-style preprocessing!** | [`PM04` (OpenMind, Wald et al., 2025)](../../curriculum/papers/modeling.md#pm04) demonstrates that benchmark wins in 3D SSL heavily depend on matched architectures, patch sizes, and fine-tuning schedules. |
| **Observation Unit & Denominators** | **Fig. 1:** $K = 70$ independent analysis teams reporting binary decisions on 9 hypotheses ($N = 108$ subjects). **Fig. 2:** $K = 64$ teams with usable unthresholded $Z$/$t$ statistical maps (6 teams excluded due to missing/corrupted/incompatible maps). | **Pretraining:** 49,561 fMRI runs segmented into variable-length SLIC spatiotemporal patches. **Downstream (Table 3):** Subject-level classification (AUROC/ACC $\uparrow$) and age regression (MAE/MSE $\downarrow$, PCC $\uparrow$) across 5 cohorts. | Always verify the exact denominator before interpreting a percentage or mean! |
| **Key Nuance / Counterexample Inside the Paper** | **Binary Decisions Disagree, Continuous Maps Agree:** Across Hypotheses #5–#9, ~20%–37% of teams reported "significant," yet Figure 2 shows a large majority core of unthresholded maps are **strongly positively correlated** (plus a small anti-correlated cluster of teams that mis-specified the parametric regressor sign!). | **Table 3 Benchmark Exception (SALD Age Regression):** While Omni-fMRI leads on ADHD-200, ABIDE, UKB Sex, and HCP-Aging, on the **SALD age regression** benchmark, competing baselines (such as BrainLM and BrainJEPA) achieve **better (lower) MAE and MSE and higher PCC** than Omni-fMRI's reported `14.37` MAE! | A single benchmark exception does not invalidate a paper's architectural contribution—but failing to notice it when claiming "universal SOTA on all tasks" is a supervisory failure. |


## Read the units and denominators before the headline

### 1. Forensic Guide to `PD01` (NARPS) Figures 1 and 2
- **Inspecting Figure 1:** Look at the bars/points showing the proportion of teams reporting a significant result for Hypotheses 1–9.
  - *Denominator check:* What is the denominator of these fractions? It is **70 analysis teams** who analyzed the same $N = 108$ sample.
  - *Epistemic trap:* Why is "33% of teams found a significant effect for Hypothesis 5" **not** equal to "there is a 33% probability that Hypothesis 5 is biologically true"? Because all 70 teams share the **exact same sampling noise** of those 108 participants! Variance across the 70 teams measures **analytical vibration (pipeline sensitivity)**, whereas biological truth also depends on population sampling variance (`PD02`) and measurement validity (`PB01`).
- **Inspecting Figure 2:** Look at the $64 \times 64$ correlation matrix of unthresholded whole-brain $Z$-maps (and its companion thresholded overlap matrix).
  - *Why did the denominator change from 70 to 64?* Locate the exact explanation in the text/caption (teams whose uploaded maps were missing, already thresholded, or spatially incompatible).
  - *Why do continuous maps agree when binary decisions disagree?* Because hard thresholding ($Z > 3.1$ vs. cluster FWE vs. ROI mask) acts as a discontinuous step function: two teams with highly correlated unthresholded maps ($r = 0.75$) that sit at $Z = 3.05$ and $Z = 3.15$ will report opposite binary conclusions ("null" vs. "significant")! Moreover, notice the small dark band of **negatively correlated maps** at the edge of the matrix—caused by teams whose GLM design matrix flipped the sign or centering of the parametric gain/loss regressor!

### 2. Forensic Guide to `PM05` (Omni-fMRI) Figure 2 and Table 3
- **Inspecting Figure 2 (Tokenization Contract):** Compare fixed parcellation tokens against Omni-fMRI's SLIC (Simple Linear Iterative Clustering) spatiotemporal patches. How does SLIC group voxels (spatial proximity + local BOLD time-series similarity)? Why does an adaptive token budget allocate smaller, denser patches to high-variance regions?
- **Inspecting Table 3 (Metric Directions & The SALD Counterexample):**
  - Trace every column header in Table 3 across the five datasets (`ADHD-200`, `ABIDE`, `UKB`, `HCP-Aging`, `SALD`).
  - Mark which metrics require **higher values ($\uparrow$)** (`AUROC`, `ACC`, `PCC` Pearson Correlation) and which require **lower values ($\downarrow$)** (`MAE` Mean Absolute Error, `MSE` Mean Squared Error).
  - Compare the `HCP-Aging` columns against the `SALD` columns. Locate one specific cell where Omni-fMRI outperforms all comparators, and locate the **SALD age-regression comparison** where Omni-fMRI (`MAE = 14.37`, `MSE = 3.29`, `PCC = 56.11`) is outperformed by competing methods (e.g., `BrainLM` and `BrainJEPA` achieving lower error and higher correlation).

## A Small Reviewer Meeting (Advocate vs. Skeptical Referee Protocol)

In top machine-learning and neuroimaging venues (ICML, NeurIPS, Nature Methods, OHBM), papers are evaluated through structured adversarial review. Conduct a 30-minute review session (either with a seminar partner or in a two-stage solo protocol):

1. **Stage 1 — The Paper's Advocate (15 minutes):**
   - Present the strongest, most charitable case for `PM05` (*Omni-fMRI*) and `PD01` (*NARPS*). What genuine limitation of fixed ROI parcellations does `Omni-fMRI`'s dynamic SLIC patching solve? What empirical improvements does it demonstrate in ablations (e.g., adaptive vs. uniform patching, dual-contrastive MAE)? Why was `NARPS` a watershed moment for open science and unthresholded map sharing (NeuroVault)?
2. **Stage 2 — The Forensic Referee (15 minutes):**
   - Challenge the split, baseline parity, preprocessing dependency, metric interpretation, or exception handling using **exact section/table locators**.
   - *Rule:* Do **not** reward an objection merely for sounding cynical ("all fMRI is noisy" or "all deep learning is a black box"). Every referee objection must point to a **concrete, verifiable vulnerability in the text** (e.g., Table 3 SALD regression underperformance, computational cost of 3D SLIC + variable-length tokens noted in Section 5, or potential motion artifacts driving "activity-aware" supervoxel boundaries) and state **what controlled experiment would test that objection**.
3. **Stage 3 — The Meta-Reviewer Decision (200 words):**
   - Write a balanced 200-word Senior Area Chair summary that synthesizes both the genuine methodological contribution and the strict boundary conditions under which the evidence holds.
